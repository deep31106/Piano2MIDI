```python
ground_truth = [
    {'pitch':60, 'onset':1.00, 'offset':1.25},
    {'pitch':60, 'onset':1.30, 'offset':1.55},
    {'pitch':64, 'onset':1.00, 'offset':1.50},
    {'pitch':67, 'onset':1.00, 'offset':1.50}
                ]

predictions = [
    {'pitch':60, 'onset':1.02, 'offset':1.27},
    {'pitch':60, 'onset':1.32, 'offset':1.57},
    {'pitch':64, 'onset':1.01, 'offset':1.52},
    {'pitch':67, 'onset':1.00, 'offset':1.50}
]
```

```python
print('Ground truth notes: ',len(ground_truth))
print('Predicted notes: ',len(predictions))

for note in ground_truth:
    print(note)
```

```python
def notes_match(true_note, predicted_note,
    onset_tolerance = 0.05,
    offset_tolerance = 0.05):

    if true_note['pitch']!=predicted_note['pitch']:
        return False

    onset_error = abs(true_note['onset']-predicted_note['onset'])

    if onset_error>onset_tolerance:
        return False

    offset_error = abs(true_note['offset']-predicted_note['offset'])

    if offset_error>offset_tolerance:
        return False

    return True
```

```python
print(notes_match(ground_truth[0], predictions[0]))

wrong_pitch = {'pitch':61, 'onset':1.02, 'offset':1.27}

print(notes_match(
    ground_truth[0],
    wrong_pitch
))

late_note = {'pitch':60, 'onset':1.20,'offset':1.27}

print(notes_match(
    ground_truth[0],
    late_note
))
```

```python
def evaluate_notes(ground_truth, predictions,
                   onset_tolerance=0.05,
                   offset_tolerance = 0.05):

    matched_true = set()
    matched_pred = set()

    for pred_idx, pred_note in enumerate(predictions):
        for true_idx, true_note in enumerate(ground_truth):

            if true_idx in matched_true:
                continue

            if notes_match(true_note, pred_note, onset_tolerance, offset_tolerance):
                matched_true.add(true_idx)
                matched_pred.add(pred_idx)
                break

    tp = len(matched_pred)
    fp = len(predictions) - tp
    fn = len(ground_truth) - tp
    precision = tp/(tp+fp) if tp+fp else 0.0
    recall = tp/(tp+fn) if tp+fn else 0.0

    f1 = (2*precision*recall/(precision + recall) if precision + recall else 0.0)

    return {
            'TP':tp,
            'FP':fp,
            'FN':fn,
            'precision':precision,
            'recall':recall,
            'F1':f1,
        }
```

```python
results = evaluate_notes(ground_truth, predictions)
print(results)
```

```python
#test A : perfect transcription
assert evaluate_notes(ground_truth, predictions)['F1']==1.0

#test B: one hallucinated note
extra_note = {'pitch':72, 'onset':2.0, 'offset':2.5}

result = evaluate_notes(ground_truth, predictions +[extra_note])
assert (result['TP'], result['FP'], result['FN'] == (4,1,0))

#test C: one missed note
result = evaluate_notes(ground_truth, predictions[:3])
assert (result['TP'], result['FP'], result['FN'] == (3,0,1))

#test D : duplicate prediction
result = evaluate_notes(ground_truth, predictions +[predictions[0]])
assert (result['TP'], result['FP'], result['FN'] == (4,1,0))

#test E : wrong pitch
wrong_pitch = [{'pitch':61, 'onset':1.02,'offset':1.27}]
result = evaluate_notes([ground_truth[0]], wrong_pitch)
assert (result['TP'], result['FP'], result['FN'] == (0,1,1))

print('All 5 evaluation tests passed')
```
