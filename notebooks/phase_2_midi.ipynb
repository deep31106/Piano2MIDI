```python
mid = mido.MidiFile(ticks_per_beat=480)
track = mido.MidiTrack()
mid.tracks.append(track)

track.append(mido.MetaMessage('set_tempo',tempo=mido.bpm2tempo(60)))

#Press sustain pedal
track.append(mido.Message('control_change', control=64, value=127, time=0))

#play C4
track.append(mido.Message('note_on', note=60, velocity=40, time=240))

#Release C4 after half beat
track.append(mido.Message('note_off',note=60, velocity=0, time=240))

#Pedal down for another beat
track.append(mido.Message('control_change', control=64, value=127, time=480))

#Release sustain
track.append(mido.Message('control_change', control=64, value=0, time=0))

mid.save('C:/Users/Deep/Desktop/Deep/Piano2MIDI/piano2midi/data/processed/sustain_test.mid')
```

```python
mid = mido.MidiFile(ticks_per_beat=480)
track = mido.MidiTrack()
mid.tracks.append(track)

track.append(mido.MetaMessage(
    "set_tempo",
    tempo=mido.bpm2tempo(60)
))

# --------------------------------------------------
# 1. Sustain pedal DOWN
# --------------------------------------------------

track.append(mido.Message(
    "control_change",
    control=64,
    value=127,
    time=0
))

# --------------------------------------------------
# 2. C major chord: C4 E4 G4
# --------------------------------------------------

track.append(mido.Message(
    "note_on",
    note=60,
    velocity=80,
    time=0
))

track.append(mido.Message(
    "note_on",
    note=64,
    velocity=75,
    time=0
))

track.append(mido.Message(
    "note_on",
    note=67,
    velocity=70,
    time=0
))

# Hold chord for 1 beat
track.append(mido.Message(
    "note_off",
    note=60,
    velocity=0,
    time=480
))

track.append(mido.Message(
    "note_off",
    note=64,
    velocity=0,
    time=0
))

track.append(mido.Message(
    "note_off",
    note=67,
    velocity=0,
    time=0
))

# --------------------------------------------------
# 3. Melody: C4 → D4 → E4
# --------------------------------------------------

# C4
track.append(mido.Message(
    "note_on",
    note=60,
    velocity=100,
    time=0
))

track.append(mido.Message(
    "note_off",
    note=60,
    velocity=0,
    time=240
))

# D4
track.append(mido.Message(
    "note_on",
    note=62,
    velocity=90,
    time=0
))

track.append(mido.Message(
    "note_off",
    note=62,
    velocity=0,
    time=240
))

# E4
track.append(mido.Message(
    "note_on",
    note=64,
    velocity=110,
    time=0
))

track.append(mido.Message(
    "note_off",
    note=64,
    velocity=0,
    time=240
))

# --------------------------------------------------
# 4. Repeated C4
# --------------------------------------------------

# First C4
track.append(mido.Message(
    "note_on",
    note=60,
    velocity=70,
    time=0
))

track.append(mido.Message(
    "note_off",
    note=60,
    velocity=0,
    time=240
))

# Second C4
track.append(mido.Message(
    "note_on",
    note=60,
    velocity=115,
    time=0
))

track.append(mido.Message(
    "note_off",
    note=60,
    velocity=0,
    time=240
))

# --------------------------------------------------
# 5. Release sustain pedal
# --------------------------------------------------

track.append(mido.Message(
    "control_change",
    control=64,
    value=0,
    time=240
))

mid.save("../data/processed/mini_piano_performance.mid")
```

```python
for message in track:
    print(message)
```
